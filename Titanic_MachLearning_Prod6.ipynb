<h1> BASIC OF MACHINE LEARNING </h1>

```python
import pandas as pd 
```

```python
Titanic = pd.read_csv('Titanic.csv')
Titanic
```

```python
# Drop irrelevant columns
Titanic = Titanic[["Survived", "Pclass", "Sex", "Age", "Embarked"]]

# Drop missing values
Titanic = Titanic.dropna()

# Encode categorical variable
Titanic["Sex"] = Titanic["Sex"].map({"male": 0, "female": 1})
Titanic["Embarked"] = Titanic["Embarked"].map({"C":0, "Q":1, "S":2})

Titanic
```

```python
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier

X = Titanic[["Pclass", "Sex", "Age","Embarked"]]
y = Titanic["Survived"]

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2)

rClassifier = RandomForestClassifier()
rClassifier.fit(X_train, y_train)
# model.fit(X,y)
```

<h1> Accuracy Test </h1>

```python
from sklearn.metrics import accuracy_score

y_pred = rClassifier.predict(X_test)
print("Accuracy:", accuracy_score(y_test, y_pred))
```

<h4> When the result of the accuracy is above 0.6 it means the model is picking up patterns it is now <strong> LEARNING </strong></h4>

<h1> TITANIC SURVAVILITY USING LOGISTIC REGRESSION </h1>

```python
# Keep only useful columns
df2 = pd.read_csv("Titanic.csv")
df2
```

```python
# Keep only useful columns
df2 = df2[["Survived", "Pclass", "Sex", "Age", "Fare", "Embarked"]]

# Drop missing values
df2 = df2.dropna()
```

```python
df2
```

```python
# Encode 'Sex' and 'Embarked'
df2["Sex"] = df2["Sex"].map({"male": 0, "female": 1})
df2["Embarked"] = df2["Embarked"].map({"S": 0, "C": 1, "Q": 2})
```

```python
df2
```

<h1>selecting X and y </h1>

```python
X = df2[["Pclass", "Sex", "Age", "Fare", "Embarked"]]
y = df2["Survived"]
```

```python
y
```

```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression

X_train, X_test, y_train, y_test = train_test_split(X, y, random_state=42)

model = LogisticRegression(max_iter=1000)
model.fit(X_train, y_train)
```

```python
from sklearn.metrics import accuracy_score

y_pred = model.predict(X_test)
print("Accuracy:", accuracy_score(y_test, y_pred))

if accuracy_score(y_test, y_pred) >=0.6:
    print("Machine is Learning")
else:
    print("Machine is not Learning")
```
